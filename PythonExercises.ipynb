## 1. Prime Number Checker - `is_prime(n)`

```python
def is_prime(n):
    if n <= 1:
        return False
    for i in range(2, int(n**0.5)+1):
        if n % i == 0:
            return False
    return True
print(is_prime(2)) 
print(is_prime(15))  
print(is_prime(17))  
```

## 2. Greatest Common Divisor (GCD) - `gcd(a, b)`

```python
def gcd(a, b):
    while b:
        a, b = b, a % b
    return a

print(gcd(12, 18))
print(gcd(100, 80))  
```

## 3. Vowel Counter in Strings - `count_vowels(s)`

```python
def count_vowels(s):
    vowels = "aeiouAEIOU"
    return sum(1 for char in s if char in vowels)
print(count_vowels("UGUR"))    
print(count_vowels("YALIN"))
```

## 4. Merging Sorted Lists - `merge_sorted_lists(list1, list2)`

```python
def merge_sorted_lists(list1, list2):
    merged = []
    i = j = 0
    while i < len(list1) and j < len(list2):
        if list1[i] < list2[j]:
            merged.append(list1[i])
            i += 1
        else:
            merged.append(list2[j])
            j += 1
    merged.extend(list1[i:])
    merged.extend(list2[j:])
    return merged
print(merge_sorted_lists([1, 3, 5], [2, 4, 6]))
```

## 5. Binary Search - `binary_search(arr, target)`

```python
def binary_search(arr, target):
    left, right = 0, len(arr) - 1
    while left <= right:
        mid = (left + right) // 2
        if arr[mid] == target:
            return mid
        elif arr[mid] < target:
            left = mid + 1
        else:
            right = mid - 1
    return -1

print(binary_search([1, 3, 5, 7, 9], 5))
print(binary_search([1, 3, 5, 7, 9], 6))  
```
